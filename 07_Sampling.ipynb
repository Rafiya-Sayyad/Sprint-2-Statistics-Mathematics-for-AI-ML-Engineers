## Sampling

### Definition
Sampling is basically the process of picking a smaller group (a sample) out of a bigger group (the population), so you can study or test something without having to collect data from literally everyone. This was briefly introduced in Notebook 1, here we go deeper into the actual methods.

### Numerical Example
Instead of surveying all 5000 employees at a company, you survey a randomly picked 200 of them, that process of picking those 200 is sampling.

### Python Implementation

```python
import random

employees = list(range(1, 5001))  # employee IDs 1 to 5000
sample = random.sample(employees, 200)

print("Sample size:", len(sample))
print("First 10 sampled employee IDs:", sample[:10])
```

#### What does this program do?
This just randomly picks 200 unique employee IDs out of the full list of 5000, simulating how you'd pick a sample from a much bigger population without surveying everyone.

### Business Example
A company doing a customer satisfaction survey can't realistically contact every single customer, so they sample a smaller group and use the results to estimate what the whole customer base probably thinks.

### AI/ML Example
Pretty much every ML dataset is a sample of some bigger population, like a dataset of 10,000 product reviews is a sample taken from literally millions of reviews that exist, and how that sample was taken affects how well a model trained on it will generalize.

### Interpretation
A sample of 200 out of 5000 is about 4% of the total population, small enough to be practical, but has to be picked carefully so it actually represents the full 5000 fairly.

#### Key Point
Sampling lets you study a smaller, manageable group instead of the entire population, as long as it's done in a way that keeps it representative.

## Random Sampling

### Definition
Random Sampling means every single member of the population has an equal chance of being picked, no bias toward any particular group or individual. It's basically the most straightforward and fair sampling method.

### Numerical Example
Picking 10 names completely at random out of a hat containing all 100 student names, every name has exactly a 1/100 chance of being picked, no favoritism.

### Python Implementation

```python
import random

students = [f"Student_{i}" for i in range(1, 101)]
random_sample = random.sample(students, 10)

print("Randomly sampled students:", random_sample)
```

#### What does this program do?
random.sample() picks 10 students completely at random from the full list of 100, where each student has an exactly equal chance of being chosen.

### Business Example
A quality control team randomly samples 10 products out of every batch of 1000 to test, since random selection means the sample isn't accidentally biased toward products from one particular shift or machine.

### AI/ML Example
When splitting a dataset into training and test sets, random sampling is often used (unless there's a reason not to) to make sure both sets are statistically similar to the full dataset, which helps the model's test performance actually reflect real-world performance.

### Interpretation
Since every student has an equal chance here, the sample of 10 should, on average, be a fair representation of the full 100 students, no particular group is more or less likely to show up.

#### Advantages
- Simple, fair, no bias in who gets picked.

#### Limitations
- Can sometimes randomly miss an important subgroup just by chance, especially with smaller sample sizes.

#### Key Point
Random Sampling gives everyone an equal chance of being picked, making it the fairest baseline sampling method.

## Stratified Sampling

### Definition
Stratified Sampling means dividing the population into subgroups (called strata) based on some characteristic, like gender or age group, and then sampling proportionally from each subgroup. This makes sure smaller subgroups aren't accidentally underrepresented, which can happen with plain random sampling.

### Numerical Example
A college has 70% undergrad students and 30% postgrad students. Instead of a plain random sample (which might randomly pick too few postgrads), stratified sampling picks 70% of the sample from undergrads and 30% from postgrads, matching the actual population ratio.

### Python Implementation

```python
import random

undergrad = [f"UG_{i}" for i in range(1, 701)]  # 700 undergrads
postgrad = [f"PG_{i}" for i in range(1, 301)]   # 300 postgrads

sample_ug = random.sample(undergrad, 70)   # 70% of a 100-person sample
sample_pg = random.sample(postgrad, 30)    # 30% of a 100-person sample

stratified_sample = sample_ug + sample_pg
print("Total sample size:", len(stratified_sample))
print("Undergrads in sample:", len(sample_ug), " Postgrads in sample:", len(sample_pg))
```

#### What does this program do?
This samples separately from each subgroup (undergrad and postgrad), keeping the same 70/30 ratio as the actual population, then combines them into one final sample.

### Business Example
A national survey about income might use stratified sampling based on state, making sure smaller states aren't underrepresented just because they have fewer people compared to larger states.

### AI/ML Example
When a dataset has imbalanced classes (like 95% 'not fraud' and 5% 'fraud' transactions), stratified sampling is used when splitting into train/test sets, to make sure the rare fraud cases still show up proportionally in both sets instead of randomly disappearing from one of them.

### Interpretation
The final sample keeps the exact 70/30 ratio of the real population, so both subgroups are fairly represented, unlike plain random sampling which could have accidentally over or under-sampled one group by chance.

#### Advantages
- Makes sure smaller/important subgroups aren't underrepresented or missed.

#### Limitations
- Requires knowing the subgroup proportions in the population upfront, which isn't always available.

#### Key Point
Stratified Sampling keeps subgroup proportions consistent between the sample and the population, which plain random sampling doesn't guarantee.

## Cluster Sampling

### Definition
Cluster Sampling means dividing the population into groups (clusters, usually based on location or natural groupings), randomly picking a few entire clusters, and then studying everyone (or a sample) within just those selected clusters, instead of sampling from the whole population directly.

### Numerical Example
A company with offices in 20 different cities wants to survey employees. Instead of randomly picking individuals from all 20 cities, they randomly pick just 4 cities (clusters) and survey all employees within those 4 cities only.

### Python Implementation

```python
import random

cities = [f"City_{i}" for i in range(1, 21)]
selected_clusters = random.sample(cities, 4)

print("Selected clusters (cities) to survey:", selected_clusters)
```

#### What does this program do?
This randomly picks 4 cities out of all 20, and in a real cluster sampling setup, every employee within those 4 selected cities would then be surveyed, not employees from the other 16 cities.

### Business Example
A retail chain testing a new store layout might pick 5 random store locations (clusters) out of 200 total stores and roll out the test fully at just those 5, instead of trying to apply partial changes across every single store.

### AI/ML Example
Cluster sampling shows up in federated learning setups, where instead of collecting data from every single device, data is grouped by device clusters (like by region or device type) and only certain clusters are used for a training round, mainly for practicality and cost reasons.

### Interpretation
Since only 4 out of 20 cities were selected, this sample is cheaper and more practical to run than surveying all 20, but it only works well if the 4 selected cities are reasonably similar to the other 16 in what you're measuring.

#### Advantages
- Much cheaper and more practical than sampling from the entire spread-out population.

#### Limitations
- If the selected clusters aren't actually representative of the others, the whole sample can end up biased.

#### Key Point
Cluster Sampling picks entire groups at random instead of individuals, trading some accuracy for a big gain in practicality and cost.

## Systematic Sampling

### Definition
Systematic Sampling means picking every kth item from a list after a random starting point, instead of picking purely randomly each time. It's simpler to carry out than full random sampling while still spreading the sample evenly across the population.

### Formula
**k = Population size / Desired sample size**

### Numerical Example
Out of 1000 customers, you want a sample of 100. k = 1000/100 = 10, so starting from a random point, you'd pick every 10th customer on the list (e.g. customer 7, 17, 27, 37...).

### Python Implementation

```python
population = list(range(1, 1001))
sample_size = 100
k = len(population) // sample_size

start = 7  # random starting point between 1 and k
systematic_sample = population[start::k]

print("Sample size:", len(systematic_sample))
print("First 10 sampled values:", systematic_sample[:10])
```

#### What does this program do?
This picks every 10th customer starting from position 7, giving an evenly spread sample of 100 customers across the full list of 1000, without having to randomly pick each one individually.

### Business Example
A factory inspecting products on an assembly line might inspect every 20th item that comes off the line, instead of randomly stopping the line to pick items, which would be way less practical in a continuous process.

### AI/ML Example
Systematic sampling is sometimes used when sampling from large streaming datasets or log files, like taking every nth record from a continuously generated server log, since it's computationally simpler than generating a fully random sample from a massive, constantly growing dataset.

### Interpretation
Since every 10th customer is picked, the sample spreads evenly across the whole customer list, assuming there's no hidden repeating pattern in the data that happens to line up with that interval of 10.

#### Advantages
- Simple and fast to apply, especially on ordered or continuously generated data.

#### Limitations
- Can go badly wrong if the population has a hidden pattern that repeats at the same interval as k, which would make the sample biased without it being obvious.

#### Key Point
Systematic Sampling picks every kth item after a random start, it's simpler than full random sampling but can break if the data has a hidden repeating pattern.

## Sampling Bias

### Definition
Sampling Bias happens when a sample is collected in a way that makes it systematically unrepresentative of the actual population, some groups end up over- or under-represented, not just by random chance, but because of how the sampling itself was done.

### Numerical Example
A survey about smartphone usage conducted only online will be biased, since it automatically excludes anyone who doesn't have internet access, the sample isn't representative of the whole population, it's skewed toward people who are already online.

### Python Implementation

```python
# Simulating biased vs unbiased sampling

population_ages = [18, 20, 22, 25, 30, 35, 40, 50, 60, 70]

# Biased sample: only picks from younger half (first 5 values)
biased_sample = population_ages[:5]

# Less biased: random sample from the whole population
import random
random_sample = random.sample(population_ages, 5)

print("Biased sample (only younger ages):", biased_sample)
print("Random sample (more representative):", random_sample)
```

#### What does this program do?
This compares deliberately picking only from the younger half of the population (biased) against a proper random sample pulled from the entire age range, showing how a biased sampling method skips large parts of the population entirely.

### Business Example
A company testing a new product only among its existing loyal customers gets biased feedback, since it completely misses the opinions of people who don't already like the brand, giving an overly positive, unrepresentative result.

### AI/ML Example
Sampling bias in ML training data directly leads to biased models, for example, a facial recognition dataset collected mostly from one demographic group will perform poorly on other demographics, since the training data never represented them fairly in the first place.

### Interpretation
The biased sample only reflects younger ages (18-30), completely missing older age groups, while the random sample has a much better chance of representing the full age spread, this is exactly the kind of gap that causes misleading conclusions.

#### Key Point
Sampling Bias means the sample systematically misrepresents the population, and it's one of the most common ways both surveys and ML models go wrong.

## Sampling Error

### Definition
Sampling Error is the natural difference between a sample's result and the actual population's true value, that happens just because you're measuring a sample instead of the whole population, even with a perfectly random, unbiased sampling method. It's unavoidable, but it tends to shrink as the sample size grows.

### Formula
**Sampling Error = Sample Statistic − Population Parameter**

### Numerical Example
The true average height of a population is 165 cm. A random sample of 50 people gives a sample average of 163.5 cm.

Sampling Error = 163.5 − 165 = −1.5 cm

### Python Implementation

```python
import numpy as np

population = np.random.normal(loc=165, scale=8, size=10000)
true_mean = np.mean(population)

sample = np.random.choice(population, size=50, replace=False)
sample_mean = np.mean(sample)

sampling_error = sample_mean - true_mean
print("True Population Mean:", round(true_mean, 2))
print("Sample Mean:", round(sample_mean, 2))
print("Sampling Error:", round(sampling_error, 2))
```

#### What does this program do?
This draws a random sample of 50 from a simulated population of 10,000 and compares the sample's mean against the true population mean, the difference between them is the sampling error, which exists purely because we only looked at a sample.

### Business Example
A political poll estimating 52% support for a candidate based on a sample has some sampling error, the true percentage across all voters might actually be 50% or 54%, which is why polls report a 'margin of error' alongside their main number.

### AI/ML Example
In ML, sampling error is part of why model performance can vary slightly between different train/test splits of the same dataset, since each split is technically a different sample, which is exactly why techniques like cross-validation are used, to average out this error across multiple samples.

### Interpretation
A sampling error of -1.5 cm means this particular sample happened to slightly underestimate the true population average. This isn't a mistake in the sampling method, it's just natural variation that comes from only measuring a subset instead of everyone.

#### Key Point
Sampling Error is the natural, unavoidable gap between a sample result and the true population value, and it tends to shrink as sample size increases.

## Sample Size

### Definition
Sample Size is just how many individuals or data points are actually included in the sample. Picking an appropriate sample size matters a lot, too small and the results are unreliable with high sampling error, too large and you're wasting time/resources beyond what's actually needed for reliable results.

### Formula
**n = (Z² × p × (1−p)) / E²**</br>
where Z = confidence level factor, p = estimated proportion, E = margin of error

### Numerical Example
For a 95% confidence level (Z=1.96), estimated proportion p=0.5 (most conservative estimate), and a margin of error E=0.05:

n = (1.96² × 0.5 × 0.5) / 0.05² ≈ 384

### Python Implementation

```python
z = 1.96  # 95% confidence level
p = 0.5   # estimated proportion (0.5 is the most conservative/safest assumption)
e = 0.05  # margin of error

sample_size = (z**2 * p * (1 - p)) / (e**2)
print("Required sample size:", round(sample_size))
```

#### What does this program do?
This uses the standard sample size formula to calculate that, for a 95% confidence level and a 5% margin of error, you'd need roughly 384 people in your sample to get reliable results, regardless of how big the actual population is.

### Business Example
A company planning a customer satisfaction survey uses this formula to figure out they need at least 384 responses to be confident their results are accurate within a 5% margin, so they know how many people to actually try to reach.

### AI/ML Example
In ML, having enough sample size (training data) directly affects how well a model can learn real patterns instead of just noise, too small a dataset often leads to overfitting, since the model ends up memorizing the limited sample instead of generalizing.

### Interpretation
A sample size of 384 is actually enough for a 95% confidence level regardless of whether the total population is 10,000 or 10 million, since this formula is based on margin of error and confidence level, not a fixed percentage of the population.

#### Advantages
- A properly calculated sample size balances reliability against the cost/time of collecting data.

#### Limitations
- Larger sample size doesn't fix sampling bias, a big biased sample is still biased, it's not automatically reliable just because it's large.

#### Key Point
Sample Size should be large enough for reliable results, but using a formula like this avoids wasting resources on an unnecessarily huge sample.

## Interview Questions

**1. What is the difference between Random Sampling and Stratified Sampling?**</br>
Random sampling gives every individual an equal chance of being picked, with no regard for subgroups. Stratified sampling deliberately divides the population into subgroups first and samples proportionally from each, making sure smaller subgroups aren't underrepresented.

**2. What is the difference between Cluster Sampling and Stratified Sampling?**</br>
Stratified sampling picks individuals from every subgroup. Cluster sampling picks entire subgroups (clusters) at random and then studies everyone within just those selected clusters, not from the ones left out.

**3. What is Sampling Bias and how is it different from Sampling Error?**</br>
Sampling bias is a systematic distortion caused by how the sample was collected, it's avoidable with a better method. Sampling error is the natural, unavoidable difference between a sample and the population that exists even with a perfectly fair sampling method.

**4. How does sample size affect sampling error?**</br>
As sample size increases, sampling error generally decreases, a bigger sample tends to be a more accurate reflection of the true population, though it doesn't fix bias, only reduces natural random variation.

**5. Why would you use Systematic Sampling instead of Random Sampling?**</br>
Systematic sampling is simpler and faster to apply, especially on ordered or continuously generated data like assembly lines or server logs, though it risks bias if the data happens to have a hidden pattern that lines up with the sampling interval.

**6. How does sampling bias affect a Machine Learning model?**</br>
If the training data was collected with sampling bias, the model learns patterns that reflect that biased sample, not the true population, which usually causes the model to perform poorly on groups or situations that were underrepresented in the training data.
