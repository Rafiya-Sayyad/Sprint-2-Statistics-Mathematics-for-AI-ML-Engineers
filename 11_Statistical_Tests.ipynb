## T-Test

### Definition
A T-Test checks whether there's a statistically significant difference between the means of two groups, it's especially useful when the sample size is small and the population standard deviation isn't known.

### Assumptions
Data should be roughly normally distributed, the two groups should be independent of each other (unless using a paired t-test), and the data should be measured on a continuous scale.

### When to Use
Use it when comparing the average of two groups, like test scores between two teaching methods, or average revenue before and after a marketing campaign, especially with smaller sample sizes.

### When Not to Use
Don't use it when comparing more than two groups at once (use ANOVA instead), or when the data clearly isn't normally distributed and the sample is too small to rely on the Central Limit Theorem (use Mann-Whitney U Test instead).

### Business Example
A company compares average customer satisfaction scores between two support teams to see if one team is genuinely performing better, or if the difference seen is just random variation.

### Python Example

```python
from scipy import stats

team_a_scores = [7, 8, 7, 9, 8, 7, 8]
team_b_scores = [6, 7, 6, 7, 6, 8, 7]

t_stat, p_value = stats.ttest_ind(team_a_scores, team_b_scores)
print("T-statistic:", round(t_stat, 3))
print("p-value:", round(p_value, 4))
```

#### What the program does
This runs an independent t-test comparing the two teams' satisfaction scores, returning a t-statistic (how big the difference is relative to the variation in the data) and a p-value (whether that difference is statistically significant).

### Interpretation
If the p-value comes out below 0.05, we'd conclude there's a statistically significant difference between the two teams' average scores, not just random noise, if it's above 0.05, we can't confidently say the teams actually differ.

## Z-Test

### Definition
A Z-Test is very similar to a t-test, it also checks for a significant difference between means, but it's used specifically when the sample size is large (usually 30+) or when the population standard deviation is actually known.

### Assumptions
Data should be roughly normally distributed (though with large samples this matters less due to the Central Limit Theorem), sample size should be large, and the population standard deviation should ideally be known.

### When to Use
Use it when working with large samples, like comparing average transaction value between two large groups of thousands of customers, where the population standard deviation is known or can be reliably estimated.

### When Not to Use
Don't use it with small sample sizes, since a t-test is more appropriate there, it accounts for the extra uncertainty that comes with smaller samples that a z-test doesn't.

### Business Example
A bank with a huge customer base compares the average account balance between customers in two different cities, using a z-test since the sample sizes are large enough.

### Python Example

```python
import numpy as np

# Simulating two large samples
np.random.seed(1)
city_a_balances = np.random.normal(5000, 800, 200)
city_b_balances = np.random.normal(5200, 800, 200)

mean_diff = np.mean(city_b_balances) - np.mean(city_a_balances)
pooled_se = np.sqrt((np.std(city_a_balances)**2 / 200) + (np.std(city_b_balances)**2 / 200))
z_score = mean_diff / pooled_se

print("Z-score:", round(z_score, 3))
```

#### What the program does
This simulates two large samples of account balances and manually calculates the z-score, which measures how many standard errors apart the two sample means are from each other.

### Interpretation
A z-score beyond about ±1.96 would typically be considered statistically significant at the common 95% confidence level, meaning the difference between the two cities' average balances likely reflects a real difference, not just random sampling variation.

## Chi-Square Test

### Definition
A Chi-Square Test checks whether there's a significant relationship between two categorical variables, by comparing the actual observed counts against what we'd expect if there was truly no relationship between them.

### Assumptions
Data should be categorical (counts/frequencies, not continuous numbers), categories should be independent of each other, and expected frequency in each category should generally be at least 5 for the test to be reliable.

### When to Use
Use it when checking if two categorical variables are related, like whether gender is associated with product preference, or whether a marketing channel is associated with customer conversion.

### When Not to Use
Don't use it on continuous numeric data directly (it needs to be categorical/counted data), and don't use it with very small expected counts in some categories, since the test becomes unreliable.

### Business Example
A retail company checks whether there's a relationship between customer age group (Young/Middle-aged/Senior) and preferred shopping channel (Online/In-store), to guide targeted marketing decisions.

### Python Example

```python
from scipy.stats import chi2_contingency
import numpy as np

# Rows: age groups, Columns: Online, In-store
observed = np.array([
    [50, 20],  # Young
    [30, 40],  # Middle-aged
    [15, 45]   # Senior
])

chi2, p_value, dof, expected = chi2_contingency(observed)
print("Chi-Square statistic:", round(chi2, 3))
print("p-value:", round(p_value, 4))
```

#### What the program does
This runs a chi-square test on a table of observed counts across age groups and shopping channels, comparing them against the counts we'd statistically expect if age group and channel preference had no relationship at all.

### Interpretation
A very small p-value here would indicate a statistically significant relationship between age group and shopping channel preference, confirming that channel preference genuinely differs across age groups rather than being random.

## ANOVA

### Definition
ANOVA (Analysis of Variance) checks whether there's a statistically significant difference between the means of THREE OR MORE groups at once, instead of running multiple separate t-tests (which would increase the chance of false positives).

### Assumptions
Data should be roughly normally distributed within each group, groups should be independent, and the variance within each group should be roughly similar (called homogeneity of variance).

### When to Use
Use it when comparing averages across 3 or more groups, like comparing average sales across 4 different store regions, or test scores across 3 different teaching methods.

### When Not to Use
Don't use it when comparing only 2 groups (a t-test is simpler and sufficient there), and don't use it when the groups have wildly different variances or aren't normally distributed, other tests handle those situations better.

### Business Example
A company compares average employee satisfaction scores across 4 different departments at once, to see if satisfaction genuinely differs by department rather than running 6 separate pairwise t-tests.

### Python Example

```python
from scipy import stats

dept_a = [7, 8, 7, 9, 8]
dept_b = [6, 7, 6, 7, 6]
dept_c = [8, 9, 8, 9, 9]
dept_d = [5, 6, 5, 6, 5]

f_stat, p_value = stats.f_oneway(dept_a, dept_b, dept_c, dept_d)
print("F-statistic:", round(f_stat, 3))
print("p-value:", round(p_value, 5))
```

#### What the program does
This runs a one-way ANOVA across all 4 departments at once, returning an F-statistic (ratio of variance between groups vs within groups) and a p-value telling us if at least one department's average is significantly different from the others.

### Interpretation
A very small p-value confirms that at least one department has a significantly different average satisfaction score, though ANOVA alone doesn't tell us WHICH specific department differs, that needs a follow-up post-hoc test.

## Mann-Whitney U Test

### Definition
The Mann-Whitney U Test is a non-parametric alternative to the t-test, it checks whether two independent groups differ, but it works by comparing ranks instead of raw values, so it doesn't require the data to be normally distributed.

### Assumptions
Data should be ordinal or continuous, the two groups should be independent, but unlike a t-test, it does NOT require the data to be normally distributed.

### When to Use
Use it when comparing two groups but the data is skewed, has outliers, or clearly isn't normally distributed, or when working with ranked/ordinal data (like survey ratings) rather than precise measurements.

### When Not to Use
Don't use it when the data genuinely is normally distributed and a t-test's assumptions are met, since the t-test is generally more statistically powerful (better at detecting a real difference) in that case.

### Business Example
A company compares customer satisfaction ratings (1-5 scale) between two product versions, using Mann-Whitney since ordinal rating data like this doesn't fit the normal distribution assumption a t-test needs.

### Python Example

```python
from scipy.stats import mannwhitneyu

product_a_ratings = [3, 4, 3, 5, 4, 3, 4]
product_b_ratings = [4, 5, 4, 5, 5, 4, 5]

u_stat, p_value = mannwhitneyu(product_a_ratings, product_b_ratings)
print("U-statistic:", u_stat)
print("p-value:", round(p_value, 4))
```

#### What the program does
This runs the Mann-Whitney U test on two sets of product ratings, which internally ranks all the combined values and checks whether one group's ranks are consistently higher or lower than the other's, without assuming normality.

### Interpretation
A small p-value here would suggest Product B's ratings are significantly different (likely higher, based on the raw numbers) than Product A's, giving a result we can trust even though rating data like this isn't normally distributed.

## Shapiro-Wilk Test

### Definition
The Shapiro-Wilk Test checks whether a dataset is likely to have come from a normally distributed population. It's often used as a first check BEFORE deciding whether to use a t-test/ANOVA (which assume normality) or a non-parametric alternative like Mann-Whitney.

### Assumptions
Works best with small to moderately sized samples, the data should be continuous, there aren't really any assumptions beyond that since this test's whole purpose is checking the normality assumption itself.

### When to Use
Use it before running a t-test, ANOVA, or any test that assumes normal data, to actually verify that assumption holds instead of just guessing.

### When Not to Use
Don't rely on it too heavily with very large sample sizes, since with a lot of data, the test can flag even tiny, practically meaningless deviations from normality as 'significant'.

### Business Example
A quality control team checks whether product weight measurements are normally distributed before deciding whether a t-test is appropriate for comparing two production batches.

### Python Example

```python
from scipy.stats import shapiro

weights = [10.1, 10.3, 9.9, 10.2, 10.0, 9.8, 10.4, 10.1, 9.9, 10.2]

stat, p_value = shapiro(weights)
print("Shapiro-Wilk statistic:", round(stat, 3))
print("p-value:", round(p_value, 4))
```

#### What the program does
This runs the Shapiro-Wilk test on a sample of product weights, checking whether this data looks like it plausibly came from a normal distribution.

### Interpretation
If the p-value is above 0.05, we fail to reject the assumption of normality, meaning it's reasonable to treat this data as normally distributed and go ahead with a t-test. If it's below 0.05, the data likely isn't normal, and a non-parametric test like Mann-Whitney would be more appropriate instead.

## Interview Questions

**1. What is the difference between a T-Test and a Z-Test?**</br>
Both check for a significant difference between two group means, but a t-test is used for smaller samples or when the population standard deviation is unknown, while a z-test is used for larger samples (usually 30+) when the population standard deviation is known.

**2. When would you use ANOVA instead of a T-Test?**</br>
ANOVA is used when comparing 3 or more groups at once. A t-test is only designed to compare 2 groups, running multiple separate t-tests instead of ANOVA increases the risk of false positives.

**3. What is the difference between a parametric test and a non-parametric test?**</br>
Parametric tests (like t-test, ANOVA, z-test) assume the data follows a specific distribution, usually normal. Non-parametric tests (like Mann-Whitney U) don't require that assumption, making them safer to use when the data is skewed or has outliers.

**4. Why would you run a Shapiro-Wilk test before a T-Test?**</br>
Because a t-test assumes the data is roughly normally distributed. Shapiro-Wilk checks whether that assumption actually holds, if it doesn't, a non-parametric alternative like Mann-Whitney U should be used instead.

**5. What does a Chi-Square Test actually measure?**</br>
It measures whether there's a significant relationship between two categorical variables, by comparing the observed counts in each category combination against what would be expected if the variables were completely unrelated.

**6. If ANOVA shows a significant result across 4 groups, does that tell you which specific groups differ?**</br>
No, ANOVA only tells you that at least one group's mean is significantly different from the others, it doesn't say which one(s). A follow-up post-hoc test (like Tukey's test) is needed to find out exactly which group(s) differ from which.
