## Notebook 10 – Hypothesis Testing

## Null Hypothesis

### Concept
The Null Hypothesis (written as H0) is basically the default assumption, that there's no effect or no difference happening. It's what we assume is true until we have strong enough evidence from the data to reject it.

### Example
If testing whether a new website design increases sign-ups, the null hypothesis would be: 'the new design has no effect on sign-up rate', essentially assuming nothing changed until proven otherwise.

### Python Implementation

```python
# Simulating sign-up rates before and after a design change
old_design_signups = [20, 22, 19, 21, 23, 20, 22]
new_design_signups = [21, 23, 20, 22, 24, 21, 23]

old_avg = sum(old_design_signups) / len(old_design_signups)
new_avg = sum(new_design_signups) / len(new_design_signups)

print("Old design average sign-ups:", round(old_avg, 2))
print("New design average sign-ups:", round(new_avg, 2))
print("Null Hypothesis assumes these averages aren't really different (any gap is just random noise).")
```

#### What the program does
This calculates the average sign-ups for both designs and prints a reminder of what the null hypothesis actually claims, that any difference we see between these two averages is just random noise, not a real effect. Whether we can actually reject that claim is determined using a statistical test (like the t-test, covered in the next notebook), not just by eyeballing the two averages.

#### Key Point
The Null Hypothesis is the 'nothing changed' assumption that a statistical test tries to find evidence against.

## Alternative Hypothesis

### Concept
The Alternative Hypothesis (written as H1 or Ha) is the opposite of the null hypothesis, it's what we're actually trying to find support for, that there IS a real effect or difference. A hypothesis test is really just checking whether the data gives us enough evidence to favor this over the null.

### Example
Continuing the website example, the alternative hypothesis would be: 'the new design does change the sign-up rate', either increasing or decreasing it.

### Python Implementation

```python
old_avg = 21.0
new_avg = 22.0

difference = new_avg - old_avg
print("Observed difference in averages:", difference)
print("Alternative Hypothesis claims this difference reflects a real effect, not just random variation.")
```

#### What the program does
This just highlights the observed difference between the two averages and frames it as what the alternative hypothesis is claiming, that this gap is a real effect, not random noise. Again, an actual statistical test is needed to confirm whether this claim holds up.

#### Key Point
The Alternative Hypothesis is the claim that there IS a real effect, and it's what we're trying to find evidence for by testing against the null.

## p-value

### Concept
The p-value tells us how likely it is to see a result as extreme as what we actually observed, purely by random chance, if the null hypothesis were actually true. A small p-value means our result would be pretty unlikely under the null hypothesis, which is evidence against it.

### Example
If a p-value comes out as 0.02, that means there's only a 2% chance of seeing a difference this big just by random luck, assuming there's actually no real effect, small enough that we'd usually reject the null hypothesis here.

### Python Implementation

```python
from scipy import stats

old_design = [20, 22, 19, 21, 23, 20, 22]
new_design = [21, 23, 20, 22, 24, 21, 23]

t_stat, p_value = stats.ttest_ind(old_design, new_design)
print("p-value:", round(p_value, 4))

if p_value < 0.05:
    print("p-value is below 0.05, reject the Null Hypothesis")
else:
    print("p-value is above 0.05, fail to reject the Null Hypothesis")
```

#### What the program does
This runs an actual t-test comparing the two sign-up groups and prints the resulting p-value, then checks it against the common 0.05 threshold to decide whether to reject the null hypothesis or not.

#### Key Point
A small p-value (typically below 0.05) means the observed result would be unlikely under the null hypothesis, giving us grounds to reject it.

## Confidence Interval

### Concept
A Confidence Interval gives a range of values that we're fairly confident the true population value actually falls within, instead of relying on just one single estimated number. A 95% confidence interval means if we repeated this sampling process many times, about 95% of the intervals we'd calculate would contain the true value.

### Example
If a survey estimates average customer satisfaction as 7.5 out of 10, with a 95% confidence interval of [7.2, 7.8], that means we're fairly confident the true average satisfaction across all customers falls somewhere in that range.

### Python Implementation

```python
import numpy as np
from scipy import stats

sample = [7, 8, 7, 9, 6, 8, 7, 8, 9, 7]

mean = np.mean(sample)
sem = stats.sem(sample)  # standard error of the mean
confidence_interval = stats.t.interval(0.95, len(sample)-1, loc=mean, scale=sem)

print("Sample Mean:", round(mean, 2))
print("95% Confidence Interval:", tuple(round(x, 2) for x in confidence_interval))
```

#### What the program does
This calculates the sample mean and then builds a 95% confidence interval around it using scipy's t-distribution interval function, giving a realistic range instead of pretending the sample mean is the exact true population value.

#### Key Point
A Confidence Interval gives a realistic range for the true population value, acknowledging that a single sample estimate is never going to be perfectly exact.

## Significance Level

### Concept
The Significance Level (written as alpha, usually 0.05) is the threshold we decide on BEFORE running a test, for how small the p-value needs to be before we're willing to reject the null hypothesis. 0.05 means we're accepting a 5% risk of wrongly rejecting a true null hypothesis.

### Example
If alpha is set to 0.05 and a test gives a p-value of 0.03, since 0.03 is less than 0.05, we reject the null hypothesis. If the p-value came out as 0.08 instead, we would NOT reject it, since it's above our chosen threshold.

### Python Implementation

```python
alpha = 0.05
p_value = 0.03

if p_value < alpha:
    print(f"p-value ({p_value}) is below significance level ({alpha}), reject Null Hypothesis")
else:
    print(f"p-value ({p_value}) is above significance level ({alpha}), fail to reject Null Hypothesis")
```

#### What the program does
This just compares a given p-value against a chosen significance level (alpha) and decides whether to reject the null hypothesis, this comparison is literally the final step of every hypothesis test.

#### Key Point
The Significance Level is the cutoff decided BEFORE testing, it's the risk level we're willing to accept for wrongly rejecting a true null hypothesis.

## Type I Error

### Concept
A Type I Error happens when we reject the null hypothesis even though it was actually true, basically a false positive. We conclude there's an effect when there really isn't one. The significance level (alpha) is literally the probability we're willing to accept of making this exact mistake.

### Example
A medical test wrongly telling a healthy person they have a disease is a Type I Error, the null hypothesis ('this person doesn't have the disease') was true, but we rejected it anyway based on a misleading test result.

### Python Implementation

```python
alpha = 0.05  # we're accepting a 5% chance of a Type I Error

# Simulating 1000 tests where the null hypothesis is actually always true
import random
random.seed(1)
false_positives = sum(1 for _ in range(1000) if random.random() < alpha)

print("Out of 1000 tests where null hypothesis is actually true,")
print("approximately", false_positives, "would incorrectly reject it (Type I Errors)")
```

#### What the program does
This simulates running 1000 hypothesis tests where the null hypothesis is actually always true, and shows that roughly 5% of them (matching our alpha of 0.05) would still incorrectly reject it, just by random chance.

#### Key Point
A Type I Error is a false positive, rejecting a null hypothesis that was actually true, and alpha directly controls how often we're willing to let this happen.

## Type II Error

### Concept
A Type II Error happens when we fail to reject the null hypothesis even though it was actually false, basically a false negative. There really IS an effect, but our test wasn't able to detect it, often because the sample size was too small or the effect was too subtle to pick up.

### Example
A medical test wrongly telling a sick person they're healthy is a Type II Error, there genuinely was a disease (the null hypothesis 'no disease' was false), but the test failed to catch it.

### Python Implementation

```python
# Smaller sample sizes make Type II errors more likely since there's less evidence to detect a real effect
small_sample = [21, 22]  # too small to reliably detect a real difference
large_sample = [21, 22, 20, 23, 21, 22, 24, 20, 23, 21]

print("Small sample size:", len(small_sample), "- higher risk of Type II Error (missing a real effect)")
print("Large sample size:", len(large_sample), "- lower risk of Type II Error")
```

#### What the program does
This just contrasts a very small sample against a larger one to illustrate the point, a small sample has less statistical power, making it more likely to miss a real effect that's actually there, which is exactly what a Type II Error is.

#### Key Point
A Type II Error is a false negative, failing to detect a real effect, and it's directly connected to having insufficient sample size or statistical power.

## One-tailed Test

### Concept
A One-tailed Test checks for an effect in only ONE specific direction, either an increase OR a decrease, not both. It's used when we specifically only care about one direction of change, which also makes it easier to reach statistical significance compared to a two-tailed test.

### Example
Testing whether a new fertilizer INCREASES crop yield (and we genuinely don't care about testing if it decreases it) is a one-tailed test, we're only checking one specific direction.

### Python Implementation

```python
from scipy import stats

control_yield = [50, 52, 51, 53, 50]
fertilizer_yield = [55, 57, 56, 58, 54]

t_stat, p_value_two_tailed = stats.ttest_ind(control_yield, fertilizer_yield)

# Converting to a one-tailed p-value (since we specifically expect an INCREASE)
p_value_one_tailed = p_value_two_tailed / 2 if t_stat < 0 else 1 - (p_value_two_tailed / 2)

print("One-tailed p-value:", round(p_value_one_tailed, 4))
```

#### What the program does
This runs a standard t-test and then converts the result into a one-tailed p-value, since we're specifically testing whether the fertilizer increases yield, not just whether it changes yield in either direction.

#### Key Point
A One-tailed Test only checks for an effect in one specific direction, which is appropriate only when we genuinely don't care about the opposite direction.

## Two-tailed Test

### Concept
A Two-tailed Test checks for an effect in EITHER direction, an increase OR a decrease, without assuming which way it might go beforehand. It's the more common, more cautious choice, since it doesn't require assuming the direction of an effect in advance.

### Example
Testing whether a new drug changes blood pressure (without assuming beforehand if it raises or lowers it) is a two-tailed test, we're checking both possible directions of change.

### Python Implementation

```python
from scipy import stats

control_group = [120, 122, 121, 119, 120]
drug_group = [115, 117, 114, 116, 113]

t_stat, p_value = stats.ttest_ind(control_group, drug_group)

print("Two-tailed p-value:", round(p_value, 4))
print("This checks for a difference in EITHER direction (higher or lower blood pressure).")
```

#### What the program does
scipy's ttest_ind() actually returns a two-tailed p-value by default, which checks for a significant difference in either direction, without assuming beforehand whether the drug group's blood pressure would be higher or lower.

#### Key Point
A Two-tailed Test is the safer default choice since it checks both directions of change, without assuming the direction of an effect ahead of time.

## Interview Questions

**1. What is the difference between Null Hypothesis and Alternative Hypothesis?**</br>
The Null Hypothesis assumes there's no real effect or difference, it's the default assumption. The Alternative Hypothesis claims there IS a real effect, and a hypothesis test is checking whether there's enough evidence in the data to reject the null in favor of the alternative.

**2. What does a p-value actually tell you?**</br>
It tells you how likely it would be to see a result as extreme as what you observed, purely by random chance, assuming the null hypothesis is true. A small p-value means your result would be unlikely under the null, which is evidence against it.

**3. What is the difference between Type I and Type II Error?**</br>
A Type I Error is a false positive, rejecting a true null hypothesis. A Type II Error is a false negative, failing to reject a false null hypothesis, basically missing a real effect that was actually there.

**4. What is the difference between a One-tailed and a Two-tailed test?**</br>
A one-tailed test only checks for an effect in one specific direction (increase or decrease, not both). A two-tailed test checks for an effect in either direction, without assuming the direction beforehand, which is why it's the more commonly used default.

**5. What does a 95% Confidence Interval actually mean?**</br>
It means if we repeated the same sampling process many times and built a confidence interval each time, about 95% of those intervals would contain the true population value. It doesn't mean there's a 95% chance the true value is in this one specific interval.

**6. Why do we set the Significance Level before running a test, not after?**</br>
Because deciding the threshold after seeing the results would let us pick whatever cutoff makes our result look significant, which defeats the whole purpose. Setting alpha beforehand keeps the test honest and prevents that kind of bias.
