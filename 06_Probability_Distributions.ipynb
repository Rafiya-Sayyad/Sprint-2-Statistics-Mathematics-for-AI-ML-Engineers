## Notebook 6 – Probability Distributions

## Bernoulli Distribution

### Definition
Bernoulli Distribution is basically the simplest distribution there is, it's for a single trial that has only 2 possible outcomes, success (1) or failure (0). Like flipping one coin, one time, that's it.

### Formula
**P(X=1) = p, P(X=0) = 1−p**</br>
where p = probability of success

### Where it is Used
It's used anywhere there's a single yes/no, pass/fail, win/lose type event. It's basically the building block that the Binomial Distribution is made from, repeating a Bernoulli trial multiple times gives you a Binomial.

### Real-world Example
Whether a single customer clicks on an ad or not is a Bernoulli trial, either they click (1) or they don't (0), there's no in-between outcome possible here.

### AI/ML Example
In ML, the output of a binary classification model (like 'is this email spam, yes or no') is modeled using a Bernoulli distribution, since it's literally a single trial with just 2 possible outcomes.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.3  # probability of success
outcomes = [0, 1]
probabilities = [1 - p, p]

plt.figure(figsize=(4,4))
plt.bar(outcomes, probabilities, color="skyblue", width=0.4)
plt.xticks([0, 1], ["Failure (0)", "Success (1)"])
plt.title("Bernoulli Distribution (p=0.3)")
plt.ylabel("Probability")
plt.show()
```

**What this shows:** This plots the two possible outcomes of a single Bernoulli trial with p=0.3, so there's a 30% bar for success and a 70% bar for failure, nothing in between since there's literally only one trial happening.

#### Key Point
Bernoulli is just one single yes/no trial, it's the simplest building block that other distributions (like Binomial) are built from.

## Binomial Distribution

### Definition
Binomial Distribution is what you get when you repeat a Bernoulli trial a fixed number of times (n trials), and count how many successes happen out of those n attempts. Each trial still has to be independent with the same success probability p.

### Formula
**P(X=k) = C(n,k) × p^k × (1−p)^(n−k)**</br>
where n = number of trials, k = number of successes, p = probability of success per trial

### Where it is Used
It's used whenever you're counting successes across repeated independent trials, like number of heads in 10 coin flips, or number of defective items in a batch of 50.

### Real-world Example
If a sales rep makes 20 calls a day and each call has a 15% chance of resulting in a sale, the number of sales they get in a day follows a Binomial Distribution.

### AI/ML Example
In A/B testing, Binomial distribution is used to model the number of conversions (like purchases) out of all the users shown a particular version of a webpage, which helps decide if one version is actually performing better than the other.

### Python Visualization

```python
import numpy as np
from scipy.stats import binom
import matplotlib.pyplot as plt

n, p = 20, 0.15
x = np.arange(0, n+1)
y = binom.pmf(x, n, p)

plt.figure(figsize=(6,4))
plt.bar(x, y, color="lightgreen")
plt.title("Binomial Distribution (n=20, p=0.15)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.show()
```

**What this shows:** This plots the probability of getting exactly 0, 1, 2...20 successes out of 20 trials with a 15% success chance each. The bars peak around 3 successes, which makes sense since 20 × 0.15 = 3, the expected number of successes.

#### Key Point
Binomial Distribution counts successes across repeated independent trials, and it's built directly from repeating the Bernoulli trial n times.

## Poisson Distribution

### Definition
Poisson Distribution models the number of times an event happens in a fixed interval of time or space, when events happen independently and at some known average rate. It's used when you're counting rare-ish events over a period, not fixed trials like Binomial.

### Formula
**P(X=k) = (λ^k × e^(−λ)) / k!**</br>
where λ = average rate of occurrence, k = number of occurrences

### Where it is Used
It's used for things like number of customer complaints per day, number of website visits per hour, number of earthquakes per year, basically counting random events over a fixed period.

### Real-world Example
A call center that receives an average of 5 calls per hour can use a Poisson distribution to estimate the probability of getting exactly 8 calls in a given hour, which helps with staffing decisions.

### AI/ML Example
Poisson distribution is used in anomaly detection systems, like flagging unusually high numbers of failed login attempts within an hour, since normal login attempt counts tend to follow a Poisson pattern around some average rate.

### Python Visualization

```python
import numpy as np
from scipy.stats import poisson
import matplotlib.pyplot as plt

lam = 5  # average rate
x = np.arange(0, 15)
y = poisson.pmf(x, lam)

plt.figure(figsize=(6,4))
plt.bar(x, y, color="salmon")
plt.title("Poisson Distribution (lambda=5)")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.show()
```

**What this shows:** This plots the probability of getting 0 through 14 calls in an hour when the average rate is 5 calls per hour. The peak is right around 5, which makes sense since that's the average, and it gets less likely the further you go from that.

#### Key Point
Poisson Distribution counts how many times something happens over a fixed time/space interval, based on a known average rate.

## Uniform Distribution

### Definition
Uniform Distribution means every outcome within a range is equally likely, there's no bias toward any particular value. This was already covered conceptually in the Data Distribution notebook, here it's revisited as one of the standard probability distributions.

### Formula
**f(x) = 1 / (b−a), for a ≤ x ≤ b**

### Where it is Used
It's used to model situations with genuinely equal chances across a range, like random number generators, or situations where literally nothing favors one outcome over another.

### Real-world Example
A raffle draw where ticket numbers 1 to 100 are all equally likely to be picked follows a uniform distribution, no number is favored over another.

### AI/ML Example
Uniform distribution is commonly used to initialize weights in a neural network before training starts, since you want the starting weights to be random but unbiased, not favoring any particular value range.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(1, 100, 1000)

plt.figure(figsize=(6,4))
plt.hist(data, bins=20, color="gold", edgecolor="black")
plt.title("Uniform Distribution (1 to 100)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

**What this shows:** This generates 1000 random values between 1 and 100 and plots them, the bars come out roughly the same height across the whole range, confirming no value range is favored over another.

#### Key Point
Uniform Distribution means every value in the range has an equal chance, it's the 'flattest' possible shape a distribution can have.

## Normal Distribution

### Definition
Normal Distribution is the classic bell-shaped curve where values cluster around the mean, and the probability drops off symmetrically the further you move away from it. This was covered in detail in the Data Distribution notebook, revisited here alongside the other standard probability distributions.

### Formula
**f(x) = (1 / (σ√(2π))) × e^(−(x−μ)² / 2σ²)**

### Where it is Used
It's used constantly across statistics and ML, basically anywhere that natural variation clusters around an average, like heights, test scores, measurement errors.

### Real-world Example
Standardized test scores (like IQ tests) are specifically designed to follow a normal distribution, with most people scoring near the average and fewer people at the extreme high or low ends.

### AI/ML Example
Many ML algorithms (Linear Regression, Logistic Regression, Gaussian Naive Bayes) either assume the data or the residuals/errors follow a roughly normal distribution, and feature scaling methods like standardization are built around this distribution's properties.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=70, scale=10, size=1000)

plt.figure(figsize=(6,4))
plt.hist(data, bins=30, color="skyblue", edgecolor="black")
plt.title("Normal Distribution (mean=70, std=10)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

**What this shows:** This generates 1000 values around a mean of 70 with a standard deviation of 10, and the histogram comes out as the classic symmetric bell shape, with most values clustering near 70.

#### Key Point
Normal Distribution is the most common and important distribution in statistics, and a lot of ML assumes data behaves this way.

## Exponential Distribution

### Definition
Exponential Distribution models the time between events in a process where events happen continuously and independently at a known average rate, basically the 'waiting time' version of the Poisson distribution. It's always skewed, with a high chance of short waits and a quickly shrinking chance of long waits.

### Formula
**f(x) = λ × e^(−λx), for x ≥ 0**</br>
where λ = rate parameter (average number of events per unit time)

### Where it is Used
It's used for modeling waiting times, like time until the next customer arrives, time until a machine part fails, or time between website visits.

### Real-world Example
If customers arrive at a store at an average rate of 10 per hour, the exponential distribution can model the actual waiting time until the next customer walks in, which is more often short, but occasionally much longer.

### AI/ML Example
Exponential distribution is used in survival analysis and reliability modeling in ML, like predicting the time until a piece of equipment fails (predictive maintenance), or how long a customer stays subscribed before churning.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.exponential(scale=1/10, size=1000)  # rate = 10 per hour

plt.figure(figsize=(6,4))
plt.hist(data, bins=30, color="plum", edgecolor="black")
plt.title("Exponential Distribution (rate=10 per hour)")
plt.xlabel("Waiting Time (hours)")
plt.ylabel("Frequency")
plt.show()
```

**What this shows:** This simulates 1000 waiting times for customers arriving at a rate of 10 per hour. The histogram is heavily right-skewed, most waiting times are short, with a long tail of occasionally longer waits, exactly the shape exponential distributions always have.

#### Key Point
Exponential Distribution models waiting time between events, and it's always right-skewed, with short waits being far more common than long ones.

## Interview Questions

**1. What's the difference between Bernoulli and Binomial Distribution?**</br>
Bernoulli is for a single trial with 2 outcomes (success/failure). Binomial is what you get when you repeat that same Bernoulli trial n times and count the total number of successes.

**2. What's the difference between Binomial and Poisson Distribution?**</br>
Binomial counts successes over a fixed number of trials (like 20 calls made). Poisson counts how many times an event happens over a fixed interval of time or space (like calls received in an hour), without a fixed "number of trials" involved.

**3. What's the relationship between Poisson and Exponential Distribution?**</br>
They're actually connected, Poisson models how many events happen in a fixed time period, while Exponential models the waiting time between those same events. They're basically two different ways of looking at the same underlying random process.

**4. Why is Exponential Distribution always right-skewed?**</br>
Because waiting times can never be negative (there's a hard floor at 0), and short waits are generally much more common than long ones, which naturally creates a long tail stretching toward longer wait times.

**5. Where would you use Uniform Distribution vs Normal Distribution in ML?**</br>
Uniform is typically used when you want unbiased randomness, like initializing neural network weights. Normal is used when you're modeling or assuming natural variation around an average, like feature distributions or model residuals.

**6. Why does Naive Bayes use Bernoulli, Binomial or Gaussian versions depending on the data?**</br>
Because the "naive" assumption changes based on what kind of feature data you're working with, Bernoulli Naive Bayes fits binary features, Multinomial/Binomial fits count data, and Gaussian Naive Bayes fits continuous, normally distributed features.
