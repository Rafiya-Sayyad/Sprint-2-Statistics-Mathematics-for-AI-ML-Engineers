## Notebook 1 – Introduction to Statistics 

### What is Statistics?

#### Step 1 – Understand the Concept
Statistics is the process of collecting, organizing, analyzing, interpreting, and presenting data.

It helps us understand data and find useful information from it.

For example, a company can collect the salaries of its employees and use statistics to find the average salary, highest salary, lowest salary, and salary differences.

Statistics is widely used in Data Science, Machine Learning, and Artificial Intelligence because these fields work with large amounts of data.

#### Step 2 – Demonstrate the Concept
#### Example

Suppose the marks of five students are:

60, 70, 75, 80, 90

Using statistics, we can calculate the average marks and understand how the students performed.

The average is:

(60 + 70 + 75 + 80 + 90) / 5 = 75

This gives us a simple summary of the students' performance.

```python
### Step 3 – Implement the Concept
marks = [60, 70, 75, 80, 90]

average = sum(marks) / len(marks)

print("Average Marks:", average)
```

##### What does this program do?
This program calculates the average marks of five students.

It demonstrates how statistics can be used to summarize a group of data values.

The result shows that the average marks are 75.

In AI/ML, similar statistical calculations are used to understand and summarize datasets before building models.

#### AI/ML Example

Suppose a Machine Learning dataset contains the ages of thousands of customers.

Statistics can be used to calculate the average age, identify unusual ages, understand the distribution of ages, and summarize the dataset.

This helps an AI/ML Engineer understand the data before training a model.

### Why Statistics is Important in AI/ML

### Step 1 – Understand the Concept

Statistics is important in AI/ML because Machine Learning models work with data.

Statistics helps us understand, summarize, analyze, and interpret data before and after building a Machine Learning model.

It helps in identifying patterns, detecting outliers, understanding relationships between variables, selecting useful features, handling missing values, and evaluating model results.

A good understanding of statistics helps an AI/ML Engineer make better decisions about data and understand how a Machine Learning model behaves.

### Step 2 – Demonstrate the Concept

#### Example

Suppose a company has customer data containing:

- Age
- Salary
- Shopping Amount
- Number of Purchases

Before building a Machine Learning model to predict customer spending, we can use statistics to:

- Calculate the average salary.
- Find the minimum and maximum shopping amount.
- Detect unusual values.
- Understand the relationship between salary and spending.
- Identify important features.

This shows how statistics helps us understand the data before using it for Machine Learning.

### Step 3 – Implement the Concept

```python
import pandas as pd

data = {
    "Age": [22, 25, 30, 35, 40],
    "Salary": [25000, 30000, 45000, 50000, 60000],
    "Shopping_Amount": [5000, 7000, 10000, 12000, 15000]
}

df = pd.DataFrame(data)

print("Average Age:", df["Age"].mean())
print("Average Salary:", df["Salary"].mean())
print("Average Shopping Amount:", df["Shopping_Amount"].mean())

print("\nCorrelation:")
print(df.corr())
```

##### What does this program do?

The program creates a small customer dataset and uses statistical methods to calculate averages and relationships between variables.

The average gives a summary of the data, while correlation helps us understand the relationship between numerical features.

These statistical techniques are commonly used during Exploratory Data Analysis (EDA) before training a Machine Learning model.

#### AI/ML Example

In a house price prediction model, an AI/ML Engineer may use statistics to understand features such as:

- House area
- Number of rooms
- Location
- Age of the house
- House price

Statistics can help identify important relationships between these features and the target variable, detect unusual values, and understand the dataset before training the model.

## Statistics vs Machine Learning
### Step 1 – Understand the Concept

Statistics and Machine Learning both work with data, but their main purposes are different.

Statistics mainly focuses on understanding data, finding relationships, making conclusions, and estimating or testing assumptions about a population.

Machine Learning mainly focuses on using data to learn patterns and make predictions or decisions on new data.

Statistics helps us understand why and how relationships exist in data, while Machine Learning focuses more on building models that can learn from data and make predictions.

Both are closely connected and are important in AI/ML.

#### Step 2 – Demonstrate the Concept

#### Example

Suppose we have data about students' study hours and exam marks.

Statistics can help us understand:

- The average study hours.
- The average marks.
- The relationship between study hours and marks.
- Whether the relationship is statistically significant.

Machine Learning can use the same data to build a model that predicts a student's marks based on the number of hours they study.

Therefore, statistics helps us understand the data, while Machine Learning can use the data to make predictions.

### Step 3 – Implement the Concept

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [45, 50, 60, 65, 75]
}

df = pd.DataFrame(data)

# Statistical analysis
print("Average Study Hours:", df["Study_Hours"].mean())
print("Average Marks:", df["Marks"].mean())
print("Correlation:", df["Study_Hours"].corr(df["Marks"]))

# Machine Learning model
X = df[["Study_Hours"]]
y = df["Marks"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[6]])

print("Predicted marks for 6 study hours:", prediction[0])
```

#### What does this program do?

The program first uses statistical methods to calculate the average values and correlation between study hours and marks.

It then uses Linear Regression to learn the relationship between study hours and marks and predicts the marks for a student who studies for 6 hours.

The statistical part helps us understand the data, while the Machine Learning part uses the data to make a prediction.

| Statistics                                     | Machine Learning                                                               |
| ---------------------------------------------- | ------------------------------------------------------------------------------ |
| Focuses on understanding data                  | Focuses on learning patterns from data                                         |
| Used for analysis and inference                | Used for prediction and decision-making                                        |
| Often emphasizes relationships and uncertainty | Often emphasizes predictive performance                                        |
| Uses methods such as hypothesis testing        | Uses algorithms such as Linear Regression, Decision Trees, and Neural Networks |
| Helps interpret data                           | Helps build predictive models                                                  |


#### AI/ML Example

In a customer churn dataset, statistics can be used to understand the average customer age, identify relationships between features, and detect unusual values.

Machine Learning can then use these features to build a model that predicts whether a customer is likely to leave the company.

Statistics and Machine Learning therefore complement each other in AI/ML projects.

## Inferential Statistics

### Step 1 – Understand the Concept

Inferential statistics is used to make conclusions or estimates about a larger population by studying a sample of that population.

It uses statistical methods such as confidence intervals and hypothesis testing to draw conclusions from sample data.

For example, instead of asking every customer in a city about their satisfaction, we can survey a sample of customers and use the results to estimate the satisfaction of the larger customer population.

### Step 2 – Demonstrate the Concept

Suppose a company has 10,000 customers.

Instead of asking all 10,000 customers whether they are satisfied, the company selects 500 customers and collects their responses.

If 420 out of the 500 customers are satisfied, the company can use inferential statistics to estimate the satisfaction level of the larger customer population.

The sample is used to draw conclusions about the population.

### Step 3 – Implement the Concept

```python
import numpy as np

sample = [72, 75, 78, 80, 76, 74, 79, 81, 77, 75]

sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)

print("Sample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
```

#### What does this program do?

The program calculates the mean and standard deviation of a sample.

The sample mean gives an estimate of the average value of the population.

The sample standard deviation shows how much the sample values vary.

In inferential statistics, sample information can be used to make estimates and conclusions about a larger population.

### AI/ML Example

Suppose an AI/ML team develops a new recommendation system.

They test the system on a sample of 1,000 users instead of all users.

The results from this sample can be analyzed using inferential statistics to estimate how the recommendation system may perform across the larger user population.

This allows teams to make decisions without collecting data from every individual.

| Descriptive Statistics                              | Inferential Statistics                               |
| --------------------------------------------------- | ---------------------------------------------------- |
| Summarizes collected data                           | Draws conclusions about a larger population          |
| Describes a dataset                                 | Estimates or tests population characteristics        |
| Uses mean, median, standard deviation, etc.         | Uses confidence intervals, hypothesis tests, etc.    |
| Does not generalize beyond the data being described | Uses a sample to make conclusions about a population |


#### Advantages
- Helps make conclusions from sample data.
- Useful when studying the entire population is difficult.
- Supports statistical decision-making.
- Useful for testing assumptions and estimating population characteristics.

#### Limitations
- Results depend on the quality of the sample.
- A biased sample can produce misleading conclusions.
- Conclusions involve uncertainty.
- Larger and more representative samples generally provide more reliable estimates.

### Population
### Step 1 – Understand the Concept

In statistics, a population is the complete group of people, objects, events, or data values that we want to study.

The population can be small or very large depending on the problem.

For example, if we want to study the salaries of all employees in a company, all the employees of that company form the population.

A population does not always mean people. It can include products, customers, transactions, students, machines, or any other group being studied.

### Step 2 – Demonstrate the Concept

#### Example

Suppose a college has 2,000 students.

If we want to study the average height of all students in the college, the 2,000 students together form the population.

Studying every student would give us information about the entire population.

```python
### Step 3 – Implement the Concept
import numpy as np

population = [160, 165, 170, 172, 168, 175, 180, 163, 169, 174]

population_mean = np.mean(population)

print("Population Mean:", population_mean)
```

#### What does this program do?
The program stores the values representing the heights of all members of a population and calculates the population mean.

The result represents the average value of the entire population.

#### AI/ML Example

Suppose an e-commerce company wants to study all transactions made on its platform during one month.

All transactions during that month can be considered the population.

An AI/ML Engineer may use this population to understand customer behavior, purchase amounts, product preferences, and other characteristics.

#### Key Point
Population = The complete group that we want to study.

#### Advantages
- Provides information about the complete group.
- Does not require estimating population characteristics from a sample.
- Gives complete information when the entire population can be studied.

#### Limitations
- Studying the entire population can be expensive.
- It can take a lot of time for large populations.
- Collecting data from every member may not be practical.

## Sample
### Step 1 – Understand the Concept

A sample is a smaller group selected from a population for the purpose of studying or analyzing the population.

When the population is very large, studying every member may be difficult, expensive, or time-consuming. In such cases, we select a representative sample and use its data for analysis.

For example, if a college has 2,000 students and we select 200 students to study their average marks, those 200 students are the sample.

### Step 2 – Demonstrate the Concept

#### Example:

A company has 10,000 customers and wants to know their satisfaction level.

Instead of asking all 10,000 customers, the company selects 500 customers and collects their feedback.

Population = 10,000 customers
Sample = 500 selected customers

The sample should represent the population as accurately as possible.

### AI/ML Example:

Suppose an ML engineer has a dataset containing 1 million customer records. They may work with a sample of the data during initial analysis or experimentation to understand patterns more quickly.

### Step 3 – Python Implementation

```python
import numpy as np

population = np.arange(1, 101)

sample = np.random.choice(population, size=10, replace=False)

print("Population:", population)
print("Sample:", sample)
```

#### What does this program do?

The program creates a population containing values from 1 to 100 and randomly selects 10 values from it without repeating any value.

The selected 10 values represent a sample of the population.

##### Interpretation

The sample contains only a small portion of the population. We can use the sample to perform analysis without working with every value in the population.

#### Business Example

A company wants to measure customer satisfaction but has millions of customers. It can randomly select a smaller number of customers and analyze their feedback.

#### Advantages
Saves time and resources.</br>
Easier to collect and analyze.</br>
Useful when studying the entire population is difficult.</br>
Allows statistical analysis on large populations.

#### Limitations
A sample may not perfectly represent the population.</br>
A biased sample can produce misleading results.</br>
The accuracy of conclusions depends on how the sample is selected.

#### Key Point
Population = Complete group being studied.

Sample = Smaller group selected from the population.

## Topic 8 – Census
### Step 1 – Understand the Concept

A census is the process of collecting information from every member of a population.

In a census, we do not select a sample. We study the complete population.

For example, if a college has 2,000 students and we collect information from all 2,000 students, it is a census.

A census provides complete information about the population, but it can require more time, money, and resources.

### Step 2 – Demonstrate the Concept

Example:

A company has 500 employees and wants to know the average age of its employees.

If the company collects the age of all 500 employees:

Population = 500 employees
Census = Information collected from all 500 employees

If the company collects the age of only 100 employees, that would be sampling, not a census.

#### Real-world example:

The Population Census of India collects demographic information about the population across the country.

### Step 3 – Python Implementation

```python
import numpy as np

population = [22, 25, 28, 30, 24, 27, 26, 29, 31, 23]

population_mean = np.mean(population)

print("Population:", population)
print("Population Mean:", population_mean)
```

##### What does this program do?

The program contains information for every member of the given population and calculates the population mean.

Since all members are included, the calculation represents the complete population rather than a sample.

##### Interpretation

The calculated mean represents the actual average of the entire population provided in the dataset.

##### Business Example

A company with 200 employees can conduct a census to collect the salary, age, department, or experience of every employee.

##### AI/ML Example

In some AI/ML projects, an organization may have complete information about all available records. Analyzing all available records can provide a complete description of that dataset.

However, when the dataset is extremely large, sampling may be used instead.

#### Advantages
Provides information about the entire population.</br>
Avoids sampling error.</br>
Gives complete population information.</br>
Useful when the population is small enough to study completely.

#### Limitations
Can be expensive.</br>
Can take a long time.</br>
Difficult to perform for very large populations.</br>
Collecting information from every member may not always be practical.

#### Key Point

Census = Data collected from every member of the population.

## Sampling
### Step 1 – Understand the Concept

Sampling is the process of selecting a smaller group of individuals or data points from a larger population for analysis.

Sampling is used when studying the entire population is difficult, expensive, or time-consuming.

The selected group is called a sample, and the complete group is called the population.

The goal is to select a sample that represents the population as accurately as possible.

### Step 2 – Demonstrate the Concept

Example:

Suppose a company has 50,000 customers and wants to know whether customers are satisfied with its service.

Instead of asking all 50,000 customers, the company selects 1,000 customers and collects their feedback.

Population = 50,000 customers
Sample = 1,000 customers
Sampling = Process of selecting the 1,000 customers

If the sample is selected properly, the company can use the sample results to understand the overall customer population.

### Step 3 – Python Implementation

```python
import numpy as np

population = np.arange(1, 101)

sample = np.random.choice(population, size=10, replace=False)

print("Population Size:", len(population))
print("Sample:", sample)
print("Sample Size:", len(sample))
```

#### What does this program do?

The program creates a population of 100 values and randomly selects 10 values from it.

The 10 selected values represent a sample of the population.

#### Interpretation

Only a small part of the population is selected for analysis. Sampling allows us to work with less data while still attempting to understand the larger population.

#### Types of Sampling

Some common sampling methods are:

- Random Sampling – Every member has a chance of being selected.
- Stratified Sampling – Population is divided into groups and samples are taken from each group.
- Cluster Sampling – Population is divided into groups and some complete groups are selected.
- Systematic Sampling – Members are selected at regular intervals.

These methods will be covered in more detail in Notebook 7 – Sampling.

#### Business Example

A company wants to evaluate a new product. Instead of asking every customer, it can select a sample of customers and collect their feedback.

AI/ML Example

An ML engineer working with millions of records may use sampling during exploratory analysis or experimentation to work with a manageable amount of data.

Sampling can also be important when creating training, validation, and test datasets.

#### Advantages
Saves time and resources.</br>
Makes large datasets easier to analyze.</br>
Useful when studying the entire population is impractical.</br>
Can provide useful estimates about a population.

#### Limitations
A poorly selected sample may not represent the population.</br>
Sampling can introduce sampling error.</br>
Biased sampling can lead to incorrect conclusions.

#### Key Point

Sampling = Selecting a smaller group from a larger population for analysis.

### Topic 11 – Features
#### Step 1 – Understand the Concept

A feature is an individual measurable property or characteristic of data that is used as an input for a Machine Learning model.

Features help a model identify patterns and make predictions.

For example, if we want to predict the price of a house, features could include:

Area of the house</br>
Number of bedrooms</br>
Number of bathrooms</br>
Location</br>
Age of the house

The value we want the model to predict is usually called the target or label.

In statistics, these characteristics can be called variables. In Machine Learning, variables used as model inputs are commonly called features.

#### Step 2 – Demonstrate the Concept

Example:

Suppose we want to predict whether a customer will purchase a product.

Age	Income	Previous Purchases	Purchased</br>
22	25000	2	No</br>
30	45000	5	Yes</br>
35	60000	8	Yes

Here:

Age → Feature</br>
Income → Feature</br>
Previous Purchases → Feature</br>
Purchased → Target/Label

The Machine Learning model uses the features to learn patterns and predict the target.

#### Step 3 – Python Implementation

```python
import pandas as pd

data = {
    "Age": [22, 30, 35],
    "Income": [25000, 45000, 60000],
    "Previous_Purchases": [2, 5, 8],
    "Purchased": ["No", "Yes", "Yes"]
}

df = pd.DataFrame(data)

features = df[["Age", "Income", "Previous_Purchases"]]
target = df["Purchased"]

print("Features:")
print(features)

print("\nTarget:")
print(target)
```

##### What does this program do?

The program separates the dataset into features and a target.

The three input columns are selected as features, while Purchased is selected as the target that we want to predict.

##### Interpretation

Features provide information that a Machine Learning model can use to learn patterns.

For example, the model may learn that customers with higher income and more previous purchases are more likely to purchase the product. This is only an example of a possible pattern; the actual relationship depends on the dataset.

Business Example

Age,

Employment status.

to predict whether a customer is likely to repay a loan.

#### AI/ML Example

In an image classification model, features can represent measurable patterns extracted from images, such as edges, shapes, textures, or learned representations.

In a house price prediction model, features can include area, bedrooms, location, and house age.

#### Advantages
Helps Machine Learning models learn from data.</br>
Makes data suitable for prediction and analysis.</br>
Useful for identifying important information in a dataset.

#### Limitations
Irrelevant features can reduce model performance.</br>
Too many features can increase model complexity.</br>
Poor-quality features can lead to inaccurate predictions.

#### Key Point

Feature = An input characteristic used by a Machine Learning model to learn patterns and make predictions.

### Topic 14 – Quantitative Data
#### Step 1 – Understand the Concept

Examples include:

#### Step 2 – Example

Business Example:

#### Step 3 – Python Implementation

```python
import numpy as np

sales = [25, 32, 18, 40, 35]

average_sales = np.mean(sales)

print("Sales:", sales)
print("Average Sales:", average_sales)
```

##### How does it work?

The program stores the number of products sold as numerical values and uses np.mean() to calculate the average.

The average sales are 30 products.

#### Advantages
Can be measured and analyzed mathematically.</br>
Useful for statistical calculations.</br>
Easy to summarize using numerical measures.</br>
Commonly used as input data for Machine Learning models.

#### Limitations
Numbers may not always explain the reason or context behind the data.</br>
Some real-world characteristics cannot be represented accurately using only numerical values.</br>
#### Key Point

Quantitative data = Numerical data that represents a quantity or amount.

### Topic 15 – Qualitative Data
#### Step 1 – Understand the Concept

Qualitative data is non-numerical data that describes qualities, characteristics, categories, or opinions.

It is used to describe what something is like rather than how much of something there is.

Examples include:

Gender,
Color,
Department,
Customer feedback,
Product category,
Marital status.

Qualitative data is usually represented using categories or descriptive labels.

#### Step 2 – Example

AI/ML Example:

Suppose a Machine Learning dataset contains customer information:

Customer	Product Type	Feedback</br>
Rafiya	Laptop	Satisfied</br>
Sonia	Phone	Neutral</br>
Divya	Tablet	Satisfied

Product Type and Feedback are qualitative data because they describe categories rather than numerical quantities.

#### Step 3 – Implementation

```python
import pandas as pd

data = {
    "Customer": ["Rafiya", "Sonia", "Divya"],
    "Product_Type": ["Laptop", "Phone", "Tablet"],
    "Feedback": ["Satisfied", "Neutral", "Satisfied"]
}

df = pd.DataFrame(data)

print(df)
print("\nFeedback Counts:")
print(df["Feedback"].value_counts())
```

##### How does it work?

The program creates a dataset containing categorical information about customers.

value_counts() counts how many times each feedback category occurs.

Here, Satisfied appears twice and Neutral appears once.

#### Advantages
Useful for understanding categories and characteristics.</br>
Helps analyze customer opinions and preferences.</br>
Can provide information that numerical data cannot directly represent.
#### Limitations
Difficult to perform direct mathematical calculations on raw qualitative data.</br>
Categories may sometimes be subjective.</br>
Qualitative data often needs to be converted into numerical form before being used by many Machine Learning algorithms.
#### Key Point

Qualitative data = Non-numerical data that describes qualities, categories, or characteristics.

### Topic 16 – Continuous Data
#### Step 1 – Understand the Concept

Continuous data is numerical data that can take any value within a given range.

It is usually obtained through measurement and can contain decimal values.

For example, height can be 165 cm, 165.5 cm, or 165.57 cm.

Other examples include:

Height,
Weight,
Temperature,
Time,
Distance.
#### Step 2 – Example

Real-World Example:

Suppose we measure the height of three students:

160.5 cm, 165.2 cm, 172.8 cm

Height is continuous data because it can have many possible values between two measurements.

#### Step 3 – Implementation

```python
heights = [160.5, 165.2, 172.8, 170.4, 168.7]

average_height = sum(heights) / len(heights)

print("Average Height:", average_height)
```

##### How does it work?

The program stores students' heights, including decimal values, and calculates their average height.

The ability to have decimal values is one of the characteristics of continuous data.

#### Advantages
Can represent measurements with high precision.</br>
Useful for statistical analysis.</br>
Common in scientific, medical, and engineering data.
#### Limitations
Measurements may contain errors.</br>
Very precise continuous data can require more storage and processing.</br>
The actual precision depends on the measuring instrument.
#### Key Point

Continuous data = Numerical data that can take any value within a range, including decimal values.

### Topic 17 – Discrete Data
#### Step 1 – Understand the Concept

Discrete data is numerical data that consists of countable and separate values.

It usually represents things that can be counted and generally takes whole-number values.

For example, the number of students in a classroom can be 30 or 31, but not 30.5.

Examples include:

Number of students,
Number of employees,
Number of products sold,
Number of calls received,
Number of cars.
#### Step 2 – Example

Business Example:

A company records the number of products purchased by customers:

2, 5, 3, 8, 1

These values are discrete because the number of products can be counted.

A customer can purchase 3 products or 4 products, but not 3.5 products.

#### Step 3 – Implementation

```python
products_sold = [2, 5, 3, 8, 1]

total_products = sum(products_sold)

print("Products Sold by Customers:", products_sold)
print("Total Products Sold:", total_products)
```

##### How does it work?

The program stores the number of products purchased by different customers and calculates the total number of products sold.

Since the values represent countable items, they are discrete data.

#### Advantages
Easy to count and record.</br>
Useful for analyzing quantities of individual items or events.</br>
Simple to summarize using counts and frequencies.
#### Limitations
Cannot represent values between two countable values.</br>
Not suitable for measurements such as height, weight, or temperature.</br>
##### Key Point

Discrete data = Countable numerical data that usually takes separate whole-number values.

### Topic 18 – Structured Data
#### Step 1 – Understand the Concept

Structured data is data that is organized in a fixed and predefined format, usually in rows and columns.

It follows a specific structure, making it easy to store, search, analyze, and process.

Examples include:

Excel spreadsheets,
SQL database tables,
Employee records,
Customer databases,
Sales records.
Step 2 – Example

#### Business Example:

A company may store employee information like this:

Employee_ID	Name	Department	Salary</br>
101	Rafiya	CSE	40000</br>
102	Sonia	ECE	45000</br>
103	Divya	CSE	50000</br>

The data is structured because every record follows the same format and each column has a specific meaning.

#### Step 3 – Implementation

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103],
    "Name": ["Rafiya", "Sonia", "Divya"],
    "Department": ["CSE", "ECE", "CSE"],
    "Salary": [40000, 45000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

##### How does it work?

The program creates a structured dataset using a Pandas DataFrame.

Each column represents a specific variable, and each row represents an employee record.

Because the data follows a predefined structure, it can easily be filtered, sorted, analyzed, and used for further processing.

#### Advantages
Easy to organize and analyze.</br>
Easy to search and filter.</br>
Works well with databases and data analysis tools.</br>
Suitable for statistical analysis and Machine Learning.
#### Limitations
Requires a predefined structure.</br>
Less flexible for information that does not follow a fixed format.</br>
Changing the structure may require modifying the database or data-processing system.
#### Key Point

Structured data = Data organized in a predefined format, usually rows and columns.

### Topic 19 – Unstructured Data
#### Step 1 – Understand the Concept

Unstructured data is data that does not follow a fixed or predefined structure such as rows and columns.

It can contain different types of information and usually requires specialized techniques to process and analyze.

Examples include:

Text documents,
Images,
Videos,
Audio recordings,
Social media posts,
Emails.

Unlike structured data, unstructured data cannot always be directly stored and analyzed using a simple table.

#### Step 2 – Example

AI/ML Example:

Suppose an online shopping company receives thousands of customer reviews such as:

"The product quality is excellent, but delivery was late."

These reviews are unstructured text data.

An AI/ML system can use Natural Language Processing (NLP) to analyze the reviews and identify whether customers have positive or negative opinions.

#### Step 3 – Implementation

A simple example of storing unstructured text in Python:

```python
customer_review = """
The product quality is excellent, but the delivery was late.
"""

print("Customer Review:")
print(customer_review)
```

##### How does it work?

The program stores a customer review as text.

The review does not follow a fixed table structure. An NLP system could later process this text to extract useful information such as sentiment, keywords, or topics.

#### Advantages
Can contain rich and detailed information.</br>
Useful for analyzing text, images, audio, and video.</br>
Provides information that may not fit into traditional tables.</br>
Widely used in AI applications such as NLP and computer vision.
#### Limitations
More difficult to process than structured data.</br>
Often requires specialized tools and Machine Learning techniques.</br>
Processing large amounts of unstructured data can require significant computational resources.
#### Key Point

Unstructured data = Data without a fixed predefined structure, such as text, images, audio, and video.`
