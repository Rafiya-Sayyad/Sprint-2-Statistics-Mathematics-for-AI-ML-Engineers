## Mean

### Definition
Mean is the average of a set of numbers. It is calculated by adding all the values together and dividing by the total number of values. It is the most commonly used measure of central tendency because it uses every value in the dataset.

### Formula
**Mean = (Sum of all values) / (Number of values)**

### Numerical Example
Suppose the marks of five students are:

60, 70, 75, 80, 90

Mean = (60 + 70 + 75 + 80 + 90) / 5 = 375 / 5 = 75

So the average mark of the class is 75.

### Python Implementation

```python
marks = [60, 70, 75, 80, 90]

mean_value = sum(marks) / len(marks)

print("Mean Marks:", mean_value)
```

#### What does this program do?
This program adds up all the marks and divides by the total number of students to calculate the mean, giving a single value that represents the overall performance of the class.

### Business Example
A retail company wants to know the average daily sales for the past week to decide staffing levels. Adding up all daily sales and dividing by 7 gives them a quick, simple benchmark to plan around.

### AI/ML Example
In a Machine Learning pipeline, the mean is used constantly during preprocessing, for example, filling in missing values in a numeric column with the column's mean (mean imputation), or in feature scaling techniques like standardization, which centers data around the mean.

### Interpretation
A mean of 75 means that, on average, students scored 75 marks. It gives a single representative number for the whole dataset, but it can be misleading if the data has outliers, since one extreme value pulls the mean toward it.

#### Advantages
- Uses every value in the dataset, so it reflects the full data.
- Easy to calculate and understand.

#### Limitations
- Gets heavily affected by outliers, one very large or very small value can pull the mean away from what most of the data actually looks like.

#### Key Point
Mean is the simplest and most common average, but it is not reliable when the data has outliers.

## Median

### Definition
Median is the middle value of a dataset when the values are arranged in ascending (or descending) order. If the number of values is odd, the median is the middle value directly. If even, it's the average of the two middle values. Unlike mean, median is not affected much by extremely large or small values.

### Formula
**If n is odd: Median = value at position (n+1)/2 after sorting**
**If n is even: Median = average of values at positions n/2 and (n/2)+1 after sorting**

### Numerical Example
Marks (sorted): 60, 70, 75, 80, 90 (n = 5, odd)

Median = value at position (5+1)/2 = 3rd value = 75

### Python Implementation

```python
marks = [90, 60, 80, 70, 75]

marks.sort()
n = len(marks)

if n % 2 == 1:
    median_value = marks[n // 2]
else:
    median_value = (marks[n // 2 - 1] + marks[n // 2]) / 2

print("Median Marks:", median_value)
```

#### What does this program do?
This program sorts the marks list and then picks the middle value, since there are an odd number of students here (5).

### Business Example
A company has the following monthly salaries (in thousands) of 5 employees: 25, 28, 30, 32, 500. One employee (maybe the CEO) earns a very high salary of 500K, an outlier. Mean would be pulled way up, but Median gives a more realistic 'typical' salary, since sorted: 25, 28, 30, 32, 500 → Median = 30.

### AI/ML Example
In real-world datasets used for ML, like house prices or income data, the median is often used instead of the mean to fill missing values or to describe a 'typical' data point, since these kinds of columns are usually skewed by a few extreme values.

### Interpretation
A median salary of 30K means half the employees earn less than 30K and half earn more, regardless of how extreme the highest or lowest salary is. This makes it a much more reliable 'typical value' when a dataset is skewed.

#### Advantages
- Not affected by outliers or extreme values.
- Gives a more realistic 'typical value' when data is skewed.

#### Limitations
- Doesn't use every value in the dataset directly, so it can ignore useful information the rest of the data carries.

#### Key Point
Median is more reliable than mean when the dataset has outliers or is skewed.

## Mode

### Definition
Mode is the value that occurs most frequently in a dataset. A dataset can have one mode (unimodal), more than one mode (multimodal), or no mode at all if every value appears the same number of times. Unlike mean and median, mode can be used with categorical (non-numeric) data too.

### Formula
**Mode = the value(s) with the highest frequency in the dataset**
(No single algebraic formula, it's identified by counting occurrences.)

### Numerical Example
Shirt sizes sold: S, M, M, L, M, XL, S, M

Counting: S=2, M=4, L=1, XL=1

Mode = M (appears most often, 4 times)

### Python Implementation

```python
from collections import Counter

sizes = ["S", "M", "M", "L", "M", "XL", "S", "M"]

counts = Counter(sizes)
mode_value = counts.most_common(1)[0][0]

print("Most sold size (Mode):", mode_value)
```

#### What does this program do?
This program counts how many times each shirt size appears, and picks the one with the highest count as the mode.

### Business Example
A clothing store owner uses mode to see that size 'M' sells the most, which directly helps decide which size to stock more of for the next order, something mean or median can't tell you since size isn't a number.

### AI/ML Example
In Machine Learning, mode is commonly used to fill missing values in categorical columns (mode imputation), since you can't calculate a mean or median for non-numeric data like 'City' or 'Gender'.

### Interpretation
A mode of 'M' means Medium is the single most common size sold. This tells the business what to prioritize, but it says nothing about how the other sizes are distributed.

#### Advantages
- Works with both numeric and categorical (non-numeric) data.
- Easy to interpret, straightforward to identify.

#### Limitations
- A dataset can have no mode, or multiple modes, which makes it less useful as a single summary value in those cases.

#### Key Point
Mode is the only central tendency measure that works well with categorical data, since it's based on frequency, not calculation.

## Weighted Mean

### Definition
Weighted Mean is similar to mean, but each value is given a different level of importance (a 'weight') based on how much it should count toward the final average, instead of every value counting equally. This is used when some values matter more than others.

### Formula
**Weighted Mean = (Sum of (value × weight)) / (Sum of weights)**

### Numerical Example
Exam 1: 70 marks, weight 20%
Exam 2: 80 marks, weight 30%
Exam 3: 90 marks, weight 50%

Weighted Mean = (70×0.20 + 80×0.30 + 90×0.50) / (0.20+0.30+0.50) = (14+24+45) / 1 = 83

### Python Implementation

```python
marks = [70, 80, 90]
weights = [0.20, 0.30, 0.50]

weighted_mean = sum(m * w for m, w in zip(marks, weights)) / sum(weights)

print("Weighted Mean:", weighted_mean)
```

#### What does this program do?
This program multiplies each exam mark by its weight, adds them up, and divides by the total weight, giving a final grade that correctly reflects how much each exam should count.

### Business Example
A company calculates an employee's final performance score using weighted mean, where 'Project Delivery' is weighted 50%, 'Teamwork' 30%, and 'Attendance' 20%, since delivery matters more to the company than attendance.

### AI/ML Example
In Machine Learning, weighted mean shows up in ensemble models, where predictions from multiple models are combined, but stronger/more accurate models are given a higher weight than weaker ones when averaging their outputs.

### Interpretation
A weighted mean of 83 (compared to a plain mean of 80) shows that giving Exam 3 more importance raised the final score, since the student did best on the exam that was weighted the heaviest.

#### Advantages
- Reflects the actual importance of each value instead of treating everything equally.

#### Limitations
- Requires the weights to be defined correctly, if the weights are wrong, the result is misleading.

#### Key Point
Weighted Mean should be used whenever some values in the dataset matter more than others, unlike a normal mean.

## Geometric Mean

### Definition
Geometric Mean is calculated by multiplying all the values together and then taking the nth root, where n is the number of values. It is mainly used when dealing with values that grow or change over time, like growth rates or percentages, rather than simple totals.

### Formula
**Geometric Mean = (x₁ × x₂ × ... × xₙ) ^ (1/n)**

### Numerical Example
Company revenue growth rate over 3 years:
Year 1: 10% growth (1.10)
Year 2: 20% growth (1.20)
Year 3: 30% growth (1.30)

Geometric Mean = (1.10 × 1.20 × 1.30) ^ (1/3)

### Python Implementation

```python
growth_rates = [1.10, 1.20, 1.30]

product = 1
for rate in growth_rates:
    product *= rate

geometric_mean = product ** (1 / len(growth_rates))

print("Geometric Mean Growth Rate:", geometric_mean)
```

#### What does this program do?
This program multiplies all the growth rates together and then takes the cube root (since there are 3 values), giving the correct average growth rate that accounts for compounding.

### Business Example
An investor wants to know the average annual return of a stock over several years where returns compound. Using a normal mean of yearly percentage returns overstates the real average, so geometric mean is used instead for an accurate figure.

### AI/ML Example
In AI/ML, geometric mean is used in some evaluation metrics that combine multiple ratios or scores fairly, for example, combining several normalized error metrics, since geometric mean handles multiplicative relationships better than a plain average.

### Interpretation
A geometric mean growth rate close to 1.198 (about 19.8%) means that if the company had grown at exactly that same rate every year, it would end up at the same final revenue as the actual varying growth rates did.

#### Advantages
- Gives an accurate average when dealing with rates, ratios, or percentage growth over time.

#### Limitations
- Cannot be used if any value in the data is zero or negative, since the calculation breaks down.

#### Key Point
Geometric Mean should be used instead of a normal mean whenever you're averaging growth rates or percentages that compound over time.

## Harmonic Mean

### Definition
Harmonic Mean is calculated by dividing the number of values by the sum of the reciprocals (1/value) of each value. It is mainly used when averaging rates or ratios, like speed, or in AI/ML, metrics like the F1-score.

### Formula
**Harmonic Mean = n / (Σ (1/x) for each value x)**

### Numerical Example
Precision = 0.80
Recall = 0.60

Harmonic Mean = 2 / (1/0.80 + 1/0.60)

### Python Implementation

```python
precision = 0.80
recall = 0.60

f1_score = 2 / ((1 / precision) + (1 / recall))

print("F1-score (Harmonic Mean):", f1_score)
```

#### What does this program do?
This program calculates the harmonic mean of precision and recall, which is exactly how the F1-score is defined in Machine Learning.

### Business Example
A logistics company calculates the average speed of a delivery vehicle that travels equal distances at different speeds (e.g. 60 km/h one way, 40 km/h the way back). A plain mean would be wrong here, harmonic mean gives the correct average speed for the full trip.

### AI/ML Example
In Machine Learning, the F1-score used to evaluate a classification model is the harmonic mean of Precision and Recall instead of a normal average, because it penalizes the score more heavily if either precision or recall is low, giving a more honest overall picture of model performance.

### Interpretation
An F1-score of about 0.686 (compared to a plain mean of 0.70) shows that harmonic mean is more sensitive to the imbalance between precision and recall, it pulls the result down more when one of the two values is weaker.

#### Advantages
- Gives a more honest average for rates/ratios, since it reduces the impact of unusually large values and reacts strongly to small ones.

#### Limitations
- Cannot be used if any value is zero, since dividing by zero is undefined. Not intuitive to interpret compared to a normal mean.

#### Key Point
Harmonic Mean is the right choice when averaging rates or ratios, and it's exactly why F1-score in ML uses it instead of a plain average of precision and recall.
