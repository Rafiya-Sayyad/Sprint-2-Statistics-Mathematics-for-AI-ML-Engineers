## Introduction to Probability

### Definition
Probability is basically just a way of measuring how likely something is to happen. It's always a number between 0 and 1, where 0 means it definitely won't happen and 1 means it definitely will. Everything in between just tells you how likely it is.

### Formula
**Probability = (Number of favorable outcomes) / (Total number of possible outcomes)**

### Numerical Example
If I flip a coin, there's 1 way to get heads out of 2 possible outcomes (heads or tails).

Probability of heads = 1/2 = 0.5

### Python Implementation

```python
favorable_outcomes = 1
total_outcomes = 2

probability = favorable_outcomes / total_outcomes
print("Probability of Heads:", probability)
```

#### What does this program do?
This just divides the number of ways heads can happen by the total number of outcomes possible, which gives 0.5, same as what we'd expect for a fair coin.

### Business Example
A delivery company might say there's a 90% probability a package arrives on time, based on how many past deliveries out of all deliveries actually made it on time.

### AI/ML Example
Probability is literally the base of everything in ML, when a model predicts something, it's usually actually outputting a probability (like 0.85 probability that an email is spam), not a hard yes/no answer.

### Interpretation
A probability of 0.5 means it's equally likely to happen or not happen, right in the middle. The closer to 1, the more likely, the closer to 0, the less likely.

#### Key Point
Probability just measures how likely something is, somewhere between 0 (impossible) and 1 (certain).

## Events

### Definition
An event is just any specific outcome or set of outcomes we're interested in. Like if we roll a die, 'getting a 4' is an event, so is 'getting an even number', which actually covers multiple outcomes at once.

### Numerical Example
Rolling a die and getting an even number is an event. The outcomes that satisfy it are 2, 4, and 6.

### Python Implementation

```python
outcomes = [1, 2, 3, 4, 5, 6]
even_event = [x for x in outcomes if x % 2 == 0]

print("Outcomes in the event 'even number':", even_event)
print("Probability of this event:", len(even_event) / len(outcomes))
```

#### What does this program do?
This filters out all the even numbers from the possible die outcomes, then calculates the probability of that event happening by dividing how many outcomes satisfy it by the total outcomes.

### Business Example
In a quality check, 'a product is defective' is an event, and the company tracks what fraction of total products checked actually fall into that event.

### AI/ML Example
In classification problems, each possible predicted class is basically an event, like 'this image is a cat' being one event out of all the possible class events the model could predict.

### Interpretation
The probability of rolling an even number is 0.5, since 3 out of the 6 total outcomes (2, 4, 6) satisfy this event.

#### Key Point
An event is just a specific outcome (or group of outcomes) we care about measuring the probability of.

## Sample Space

### Definition
Sample Space is literally just the list of every possible outcome of something. For a die roll it's {1,2,3,4,5,6}, for a coin flip it's {heads, tails}. Every event we talked about above is basically a smaller piece taken from this sample space.

### Numerical Example
For a single coin flip, the Sample Space is: {Heads, Tails}

For rolling a die, the Sample Space is: {1, 2, 3, 4, 5, 6}

### Python Implementation

```python
sample_space_die = set(range(1, 7))
sample_space_coin = {"Heads", "Tails"}

print("Sample Space (Die):", sample_space_die)
print("Sample Space (Coin):", sample_space_coin)
```

#### What does this program do?
This just defines both sample spaces using Python sets, since a sample space is really just a collection of all the unique possible outcomes.

### Business Example
A call center defines its sample space for a customer call outcome as {Resolved, Escalated, Abandoned}, every single call has to fall into exactly one of these.

### AI/ML Example
When building a classifier, the sample space is basically the full list of classes the model can possibly predict, like {spam, not spam} for an email classifier.

### Interpretation
The sample space just tells us everything that could possibly happen, nothing more, nothing less, it's the full universe of outcomes we're working with.

#### Key Point
Sample Space is the complete list of every possible outcome, and every event is just a subset of it.

## Independent Events

### Definition
Two events are independent if one happening has literally zero effect on whether the other happens. Like flipping a coin twice, what you got on the first flip doesn't change the odds for the second flip at all.

### Formula
**P(A and B) = P(A) × P(B), only true if A and B are independent**

### Numerical Example
Flipping a coin twice. Probability of heads then heads again:

P(Heads, Heads) = 0.5 × 0.5 = 0.25

### Python Implementation

```python
p_heads = 0.5

p_heads_twice = p_heads * p_heads
print("Probability of Heads then Heads:", p_heads_twice)
```

#### What does this program do?
Since the two coin flips don't affect each other at all, we can just multiply their individual probabilities directly to get the combined probability.

### Business Example
A factory checks two separate machines that work completely independently of each other, so the probability both machines break down on the same day is just the product of each machine's individual breakdown probability.

### AI/ML Example
In Naive Bayes classifiers, the model actually assumes all the features are independent of each other (hence 'naive'), which isn't always true in real life, but it makes the math way simpler and still works surprisingly well in practice.

### Interpretation
A 0.25 probability means getting heads twice in a row happens about 1 in 4 times, which makes sense since each flip on its own is already a coin toss.

#### Key Point
Independent Events don't affect each other at all, so you just multiply their individual probabilities together.

## Dependent Events

### Definition
Dependent events are the opposite, one event happening actually changes the probability of the other one happening. Classic example is drawing cards without putting them back, since removing a card changes what's left in the deck.

### Formula
**P(A and B) = P(A) × P(B given A)**

### Numerical Example
Drawing 2 aces in a row from a 52-card deck, without replacing the first card:

P(1st Ace) = 4/52
P(2nd Ace given 1st was Ace) = 3/51 (one ace and one card are already gone)

P(both Aces) = (4/52) × (3/51)

### Python Implementation

```python
p_first_ace = 4/52
p_second_ace_given_first = 3/51

p_both_aces = p_first_ace * p_second_ace_given_first
print("Probability of drawing 2 Aces in a row:", round(p_both_aces, 4))
```

#### What does this program do?
This calculates the probability of the first ace, then the probability of the second ace given the first one is already gone, and multiplies them, since the two draws actually depend on each other here.

### Business Example
An HR team analyzing promotions might find that 'being promoted' and 'having completed a specific training' are dependent events, since completing the training actually changes the likelihood of getting promoted.

### AI/ML Example
In sequence models (like predicting the next word in a sentence), each word's probability actually depends on the words that came before it, that's literally dependent events in action, the model isn't treating every word as independent.

### Interpretation
The probability of drawing two aces in a row comes out pretty small (around 0.45%), which makes sense since aces are rare, and it gets even rarer when one is already removed from the deck for the second draw.

#### Key Point
Dependent Events affect each other, so the probability of the second event changes based on whether the first one already happened.

## Conditional Probability

### Definition
Conditional Probability is the probability of something happening given that something else has already happened. It's written as P(A|B), which reads as 'probability of A given B'. This is directly connected to dependent events from above.

### Formula
**P(A|B) = P(A and B) / P(B)**

### Numerical Example
Out of 100 students, 40 study Python, 30 study SQL, and 20 study both.

P(studies SQL | studies Python) = P(both) / P(Python) = 20/100 / 40/100 = 0.5

### Python Implementation

```python
p_both = 20/100
p_python = 40/100

p_sql_given_python = p_both / p_python
print("P(studies SQL | studies Python):", p_sql_given_python)
```

#### What does this program do?
This divides the probability of studying both subjects by the probability of studying Python alone, giving us the chance someone studies SQL specifically among the group that already studies Python.

### Business Example
A marketing team calculates the probability a customer buys Product B given that they already bought Product A, which directly helps decide who to target with a 'customers also bought' recommendation.

### AI/ML Example
Conditional probability is basically the backbone of recommendation systems, predicting P(user likes Movie B | user liked Movie A) is exactly how a lot of 'recommended for you' logic works under the hood.

### Interpretation
A 0.5 probability means that among students who study Python, exactly half of them also study SQL, which is actually higher than the overall 30% of all students who study SQL, showing some connection between the two subjects.

#### Key Point
Conditional Probability tells you the chance of something happening, specifically within a situation where something else is already known to be true.

## Bayes' Theorem

### Definition
Bayes' Theorem is basically a formula that lets you flip a conditional probability around, so you can go from P(B|A) to P(A|B) instead, which is super useful when the thing you actually want to know (P(A|B)) is harder to measure directly than the reverse.

### Formula
**P(A|B) = (P(B|A) × P(A)) / P(B)**

### Numerical Example
1% of people have a certain disease. A test is 90% accurate for people who actually have it (true positive), and has a 5% false positive rate for people who don't.

If someone tests positive, what's the actual probability they have the disease?

P(Disease|Positive) = (0.90 × 0.01) / [(0.90×0.01) + (0.05×0.99)]

### Python Implementation

```python
p_disease = 0.01
p_positive_given_disease = 0.90
p_positive_given_no_disease = 0.05
p_no_disease = 1 - p_disease

p_positive = (p_positive_given_disease * p_disease) + (p_positive_given_no_disease * p_no_disease)
p_disease_given_positive = (p_positive_given_disease * p_disease) / p_positive

print("P(Disease | Positive Test):", round(p_disease_given_positive, 4))
```

#### What does this program do?
This calculates the overall probability of testing positive first (combining both the true positive and false positive paths), then uses Bayes' formula to flip it around and find the actual probability someone has the disease given they tested positive.

### Business Example
An insurance company uses Bayes' Theorem to update the estimated risk of a customer filing a claim, starting with a general base rate, and then updating it as new information comes in, like their driving history.

### AI/ML Example
Naive Bayes classifiers (used for things like spam detection) are literally built directly on this formula, they calculate P(spam | words in email) using Bayes' Theorem, starting from how often those words appear in spam vs non-spam emails.

### Interpretation
The actual probability of having the disease given a positive test comes out way lower than 90%, something like 15-16%, which surprises a lot of people, but it makes sense because the disease itself is so rare (1%) that even a small false positive rate causes a lot of false alarms.

#### Advantages
- Lets you update a belief/probability as new evidence comes in, instead of only relying on the original assumption.

#### Limitations
- Needs accurate prior probabilities to begin with, if the starting assumptions are wrong, the final answer will be wrong too.

#### Key Point
Bayes' Theorem flips a conditional probability around, and it's exactly why a positive medical test for a rare disease doesn't mean you're actually likely to have it.

## Probability Rules

### Definition
There's a few basic rules that probability always follows. The Addition Rule is for 'or' situations (P(A or B) = P(A) + P(B) − P(A and B)), and the Multiplication Rule is for 'and' situations (P(A and B) = P(A) × P(B|A)). There's also the Complement Rule, P(not A) = 1 − P(A).

### Formula
**Addition: P(A or B) = P(A) + P(B) − P(A and B)**
**Complement: P(not A) = 1 − P(A)**

### Numerical Example
Rolling a die, what's the probability of getting a 3 OR an even number?

P(3) = 1/6, P(even) = 3/6, P(3 and even) = 0 (3 isn't even, no overlap)

P(3 or even) = 1/6 + 3/6 − 0 = 4/6

### Python Implementation

```python
p_3 = 1/6
p_even = 3/6
p_3_and_even = 0  # no overlap, 3 isn't even

p_3_or_even = p_3 + p_even - p_3_and_even
print("P(3 or Even):", round(p_3_or_even, 4))

# Complement rule example
p_even_complement = 1 - p_even
print("P(not even):", p_even_complement)
```

#### What does this program do?
This applies the addition rule directly, adding the two individual probabilities and subtracting any overlap (which is 0 here since 3 can't also be even), and separately shows the complement rule for 'not even'.

### Business Example
A store calculates the probability a customer buys a shirt OR a jacket using the addition rule, since some customers might buy both, that overlap has to be subtracted so it isn't counted twice.

### AI/ML Example
These rules show up constantly when combining multiple model predictions or conditions, like calculating the probability that at least one of several independent model checks flags an issue, which uses the complement rule (1 minus the probability that none of them flag it).

### Interpretation
P(3 or even) comes out to 4/6, which makes sense since the only numbers that satisfy neither condition are 1 and 5, leaving 4 out of 6 numbers that do.

#### Key Point
These rules (addition, multiplication, complement) are the basic toolkit for combining probabilities of different events together correctly.

## Random Variables

### Definition
A random variable is just a variable whose value depends on the outcome of some random event. It can be discrete (specific countable values, like number of heads in 3 flips) or continuous (any value in a range, like someone's height).

### Numerical Example
Let X = number of heads when flipping a coin 2 times.

X can be 0, 1, or 2, depending on what actually happens when you flip.

### Python Implementation

```python
import random

def flip_two_coins():
    flips = [random.choice(["H", "T"]) for _ in range(2)]
    return flips.count("H")

results = [flip_two_coins() for _ in range(10)]
print("Number of heads in each trial:", results)
```

#### What does this program do?
This simulates flipping 2 coins, 10 separate times, and counts how many heads showed up each time. Each one of those counts is literally a value the random variable X could take.

### Business Example
An insurance company treats 'number of claims filed by a customer in a year' as a random variable, since it's unknown in advance but follows some underlying probability pattern they can study.

### AI/ML Example
In ML, model outputs are often treated as random variables, like the predicted class probability, since the exact value depends on the randomness in the data and sometimes the randomness in the model's own training process.

### Interpretation
Each run gives a different sequence of head counts (0, 1, or 2) since it's random, but if you ran this thousands of times, the overall pattern of how often each value shows up would start to stabilize into a predictable distribution.

#### Key Point
A Random Variable is just a way of representing an outcome that depends on randomness, and it can be discrete or continuous depending on what it's measuring.

## Expected Value

### Definition
Expected Value is basically the long-run average outcome you'd expect if you repeated something many, many times. It's calculated by multiplying each possible outcome by its probability and adding them all up.

### Formula
**E(X) = Σ (x × P(x)) for all possible values of x**

A game costs $5 to play. You  win $20 with 10% probability, or win $0 otherwise.

Expected Value = (20 × 0.10) + (0 × 0.90) = 2 + 0 = 2

Since it costs $5 to play but the expected winnings is only $2, this game isn't worth playing long-term.

### Python Implementation

```python
outcomes = [20, 0]
probabilities = [0.10, 0.90]

expected_value = sum(x * p for x, p in zip(outcomes, probabilities))
print("Expected Value of the game:", expected_value)
```

#### What does this program do?
This multiplies each possible winning amount by its probability of happening, and adds them together, giving the average amount you'd expect to win per game if you played it a huge number of times.

### Business Example
An insurance company calculates the expected payout per policy (claim amount × probability of a claim) to decide how much to charge in premiums, so they come out profitable on average across all their customers.

### AI/ML Example
Expected value is central to reinforcement learning, where an agent is literally trying to maximize the expected future reward of its actions, not just the reward from one single action.

### Interpretation
An expected value of $2 means that if you played this game a huge number of times, you'd average $2 in winnings per game, which is less than the $5 it costs to play, so on average you'd actually lose $3 per game long-term.

#### Key Point
Expected Value tells you the long-run average outcome, which is why it's the key number used to judge if a bet, game, or decision is actually worth it over time.

## Interview Questions

**1. What is the difference between Independent and Dependent events?**</br>
Independent events don't affect each other at all, like two separate coin flips. Dependent events do affect each other, like drawing cards without putting them back, since the first draw changes what's left for the second draw.

**2. What is Conditional Probability?**</br>
It's the probability of an event happening given that another event has already happened, written as P(A|B). It's basically asking "given this situation, what's the chance of this other thing happening?"

**3. What is Bayes' Theorem used for?**</br>
It lets you flip a conditional probability around, going from P(B|A) to P(A|B). It's useful when the probability you actually want to know is harder to measure directly than its reverse, like going from test accuracy to actual disease probability.

**4. What's the difference between Sample Space and an Event?**</br>
Sample Space is literally every possible outcome of something. An Event is just a specific outcome or group of outcomes we're interested in, basically a smaller piece taken out of the sample space.

**5. What is Expected Value and why does it matter?**</br>
Expected Value is the long-run average outcome you'd expect if something was repeated many times. It matters because it's how you judge whether a decision, bet, or game is actually worth it over the long run, not just based on one single try.

**6. How is Bayes' Theorem connected to Naive Bayes in Machine Learning?**</br>
Naive Bayes classifiers literally use Bayes' Theorem directly to calculate the probability of a class given the input features, it's called "naive" because it assumes all the features are independent of each other, which simplifies the math a lot.
